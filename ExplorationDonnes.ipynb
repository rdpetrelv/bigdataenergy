## Lecture des données

```python
# Path to wd, chaneg to your own directory
path = f"C:/Users/HP/Documents/07. EMSE/Defi Big Data/Exploration des données/Defi1"

dataframes = {}

for file_name in os.listdir(path):
    if file_name.endswith('.csv'):
        file_path = os.path.join(path, file_name)
        df = pd.read_csv(file_path, delimiter=";")
        dataframes[file_name] = df
```
